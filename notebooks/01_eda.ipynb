## 3. Data quality

### 3.1 Missing values

```python
missing = (
    df.isnull().sum().to_frame("missing_count").sort_values("missing_count", ascending=False)
)
missing[missing["missing_count"] > 0]
```

```python
cols_with_na = missing.index[missing["missing_count"] > 0].tolist()

rows = []
for c in cols_with_na:
    rate = df.groupby(df[c].isna())[TARGET].mean().mul(100)
    rows.append({
        "column": c,
        "churn_% when present": rate.get(False, np.nan),
        "churn_% when missing": rate.get(True, np.nan),
    })
pd.DataFrame(rows).set_index("column")
```

> **Note (Critical Finding — Informative Missingness):**
- Three columns have missing values: `DaySinceLastOrder` (**213**, 5.4%), `Tenure` (**194**, 4.9%), and `WarehouseToHome` (**169**, 4.3%).
- Missingness in `Tenure` and `WarehouseToHome` is **not missing at random (MNAR)**: customers with missing `Tenure` churn at **32.99%** (vs. **16.28%** when present), and those with missing `WarehouseToHome` churn at **35.50%** (vs. **16.28%** when present) — more than **double** the baseline rate.
- **Modeling implication:** Do not simply drop these rows; create binary missingness indicator flags (`Tenure_isna`, `WarehouseToHome_isna`) prior to median imputation.

### 3.4 Outliers

```python
num_cols = ["Tenure", "WarehouseToHome", "DaySinceLastOrder", "CashbackAmount"]
discrete_cols = ["NumberOfDeviceRegistered", "NumberOfAddress", "SatisfactionScore", "Complain"]

def iqr_outliers(s):
    s = s.dropna()
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    return int(((s < q1 - 1.5 * iqr) | (s > q3 + 1.5 * iqr)).sum())

pd.DataFrame({
    "outliers": {c: iqr_outliers(df[c]) for c in num_cols + ["NumberOfDeviceRegistered", "NumberOfAddress"]},
}).assign(pct=lambda d: d["outliers"] / len(df) * 100)
```

> **Note:**
- **`CashbackAmount`** (316 outliers, **8.02%**) and **`NumberOfDeviceRegistered`** (271 outliers, **6.88%**) show the highest counts under the standard 1.5×IQR rule. For `NumberOfDeviceRegistered`, this is a mathematical artifact of a tight IQR ($Q_1=3, Q_3=4$) rather than true data corruption.
- **`DaySinceLastOrder`** has **43 outliers (1.09%)** representing long-inactive buyers (up to 46 days).
- **`Tenure`** (4 outliers), **`NumberOfAddress`** (3 outliers, max 22), and **`WarehouseToHome`** (1 extreme outlier at 127) have isolated extreme values that can be capped (winsorized) if using linear models.

- The target exhibits **moderate class imbalance (~5:1 ratio)**. Downstream modeling should use **stratified train/test splits**, class-weighted loss functions (`class_weight='balanced'`), and precision-recall / ROC-AUC metrics rather than raw accuracy.

### 6.2 Churn rate by category

```python
def churn_rate_by(col, ax, order=None):
    rate = df.groupby(col)[TARGET].mean().mul(100)
    if order is not None:
        rate = rate.reindex(order)
    sns.barplot(x=rate.index.astype(str), y=rate.values, ax=ax)
    ax.axhline(df[TARGET].mean() * 100, color="red", ls="--", lw=1, label="overall rate")
    ax.set(title=f"Churn rate (%) by {col}", xlabel="", ylabel="Churn %")
    ax.tick_params(axis="x", rotation=25)
    ax.legend()

fig, axes = plt.subplots(3, 2, figsize=(13, 12))
for ax, c in zip(axes.ravel(), cat_like):
    churn_rate_by(c, ax)
plt.tight_layout()
plt.savefig(FIGURES_DIR / "06_churn_rate_by_categories.png", dpi=300, bbox_inches="tight")
plt.show()
```

#### Exact Churn Rates by Key Categorical Drivers

```python
print(df.groupby("Complain")[TARGET].mean().mul(100).round(1))
print(df.groupby("SatisfactionScore")[TARGET].mean().mul(100).round(1))
print(df.groupby("PreferedOrderCat")[TARGET].mean().mul(100).round(1))
```

> **Note:**

This suggests either an inverted scale in survey collection or that high-scoring users who encounter a sudden service failure leave abruptly.
- **`MaritalStatus`:** **`Single`** customers churn at the highest rate (~26%+), well above `Divorced` (~15%) and `Married` (~12%).

### 6.3 Churn rate by tenure band

```python
bins = [-1, 3, 6, 12, 24, np.inf]
labels = ["0-3 mo", "4-6 mo", "7-12 mo", "13-24 mo", "25+ mo"]
tenure_band = pd.cut(df["Tenure"], bins=bins, labels=labels)

band = (
    df.groupby(tenure_band, observed=True)[TARGET]
      .agg(customers="count", churn_rate="mean")
      .assign(churn_rate=lambda d: d["churn_rate"] * 100)
)
display(band)

fig, ax = plt.subplots(figsize=(7, 4))
sns.barplot(x=band.index.astype(str), y=band["churn_rate"], ax=ax)
ax.set(title="Churn rate (%) by tenure band", xlabel="Tenure", ylabel="Churn %")
plt.savefig(FIGURES_DIR / "07_churn_rate_by_tenure_band.png", dpi=300, bbox_inches="tight")
plt.show()
```

> **Note:**
- **The First-90-Days Cliff (`0-3 mo`):** Churn is overwhelmingly concentrated in the first 3 months at **43.16%** (out of 1,089 customers in this band, ~470 churned — accounting for **~70% of all churn in the dataset**).
- **Rapid Stabilization (`4-24 mo`):** Once a customer survives past month 3, churn drops more than **6-fold** to **6.98%** (`4-6 mo`), **5.29%** (`7-12 mo`), and **5.99%** (`13-24 mo`).
- **Long-Term Loyalty (`25+ mo`):** Customers with over 2 years of tenure have a **0.00% churn rate**.

## 8. Key Insights & Recommendations

### 8.1 Key Insights
1. **Severe Early-Life Churn (The 0–3 Month Danger Zone):**
   - Overall churn is **17.1%**, but **43.16%** of customers with **0–3 months of tenure** churn (median tenure for churners is **1 month** vs. **10 months** for retained users). Once customers pass the 3-month mark, churn drops below **7%**, and reaches **0%** after 25 months.
2. **Unresolved Complaints Triple Churn Risk:**
   - **28%** of customers logged a complaint, and their churn rate jumps to **32.8%** compared to **10.9%** for non-complainers ($3\times$ higher risk). When combined with `SatisfactionScore = 5`, complaint churn reaches **40.8%**.
3. **Mobile Category Vulnerability & Label Duplication:**
   - Shoppers in **`Mobile Phone` (28.7% churn)** and **`Mobile` (28.4% churn)** leave at nearly **$3\times$ the rate** of `Laptop & Accessory` (**9.7%**) and **$6.5\times$ the rate** of `Grocery` (**4.4%**). Additionally, `Mobile` and `Mobile Phone` are duplicate labels for the same category.
4. **Cashback & Delivery Distance Effects:**
   - Retained users earn higher median cashback (**166.08** vs. **149.34**) and live closer to warehouses (median **13.0** vs. **15.0**), indicating that loyalty rewards and faster fulfillment support retention.
5. **Critical Data Quality Findings for Modeling:**
   - **Informative missingness:** Missing values in `Tenure` (**32.99% churn**) and `WarehouseToHome` (**35.50% churn**) carry strong predictive signal (more than double the ~16.3% churn rate when present).
   - **Duplicates:** **671 rows (17.03%)** are exact duplicates that must be handled carefully to avoid train-test leakage.

---

### 8.2 Actionable Recommendations

#### A. Business & Retention Strategy
- **Launch a First-90-Days Onboarding & Retention Program:** Trigger automated post-purchase check-ins, second-order discount vouchers, or bonus cashback milestones specifically for customers in their **first 1–3 months** to bridge them past the 3-month cliff.
- **Fast-Track Complaint Resolution (SLA Escalation):** Treat any logged complaint (`Complain = 1`) as a high-priority churn alert. Route complaining customers—especially new users and high-score buyers—to a dedicated recovery workflow with proactive follow-up.
- **Audit the Mobile & Mobile Phone Category Experience:** Investigate product quality, seller reliability, return rates, and pricing in the `Mobile` / `Mobile Phone` vertical to address why nearly 1 in 3 buyers churns.
- **Optimize Cashback Structure & Fulfillment SLAs:** Use targeted cashback incentives to lock in single/multi-device shoppers, and set clear delivery expectations for customers with high `WarehouseToHome` distances.

#### B. Data Preprocessing & Feature Engineering Pipeline
- **Consolidate Categories:** Merge `"Mobile"` and `"Mobile Phone"` in `PreferedOrderCat` into a single `"Mobile Phone"` category.
- **Preserve Missingness Signal:** Add binary indicator columns (`Tenure_isna`, `WarehouseToHome_isna`, `DaySinceLastOrder_isna`) *before* imputing missing numeric values with medians.
- **Engineer High-Signal Features:** Create a `Tenure_Band` / `Is_New_Customer` flag (`Tenure <= 3`), interaction terms (`Complain × SatisfactionScore`), and `Cashback_per_Tenure` ratios.
- **Handle Duplicates & Class Imbalance:** Deduplicate or group identical records prior to splitting so identical rows do not leak across train and test sets, and train models using **stratified k-fold cross-validation** with class weighting (`class_weight='balanced'` / `scale_pos_weight`).
